```python

class Jar:
    def __init__(self):
        self.candies = 100


class Withdraw:
    def execute(self, jar, amount):
        if amount <= jar.candies:
            jar.candies -= amount
            print("Withdraw:", amount)
        else:
            print("Not enough candies!")

        print("Remaining candies:", jar.candies)


class Deposit:
    def execute(self, jar, amount):
        jar.candies += amount
        print("Deposit:", amount)
        print("Total candies:", jar.candies)


jar = Jar()

user1 = Withdraw()
user2 = Deposit()
user3 = Withdraw()
user4 = Deposit()

amount = int(input("User 1: Enter candies to withdraw: "))
user1.execute(jar, amount)

amount = int(input("User 2: Enter candies to deposit: "))
user2.execute(jar, amount)

amount = int(input("User 3: Enter candies to withdraw: "))
user3.execute(jar, amount)

amount = int(input("User 4: Enter candies to deposit: "))
user4.execute(jar, amount)
```

```python
## Obj only access one class
```

```python
## Abstraction is focusing on the essential feature of an obj
```

```python
## Encapsulation Define
```

```python
from abc import ABC, abstractmethod
class Payment(ABC):

    def process_payment(self,amount):
        print(amount)
#--------------------------------------
class CreditCard(Payment):

    def receipt(self,amount):
        print(amount)
```

```python
Py_1 = Payment()
Py_1.process_payment(10000) #parent object accesssing its own's class method
```

```python
Child1 = CreditCard()
Child1.process_payment(100) # # Child Object Accessing the parent's class method
```

```python
from abc import ABC, abstractmethod
# 1. Define the abstarct Base Class (The BluePrint)
class Payment(ABC):

    @abstractmethod
    def process_payment(self, amount):
        """The method must be implemented by all subclasses"""
        pass
    @abstractmethod
    def receipt(self, amount):
        """Regular methods can still exist in an abstract class"""
        print(f"Receipt generated for ₹{amount}")

# 2.Implement Concrete Subclasses
class CreditCard(Payment):
    def process_payment(self, amount):
        print(f"Processsing Credit Card payment of ₹{amount} via Secure Gateway.")
    def receipt(self, amount):
        pass
class UPI(Payment):
    pass
    def process_payment(self, amount):
        print(f"Processsing UPI payment of ₹{amount} via Scan & Pay.")
    def receipt(self,amount):
        pass

cc = CreditCard()
cc. process_payment(5000)
cc. receipt(5000)

upi = UPI()
upi.process_payment(1200)
```

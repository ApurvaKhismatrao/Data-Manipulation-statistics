# Concatenation of Arrays

```python
a1 = np.ones((3,5), dtype = int)
a2 = np.zeros((3,2), dtype = int)
print(a1)
print(a2)

A3 = np.concatenate((a1,a2), axis=1)    # axis= 1 for rows & axis=0 for column
A3
```

```python
a6 = np.ones((4,3), dtype = int)
a7 = np.zeros((3,2), dtype = int)
print(a6)
print(a7)

A8 = np.concatenate((a6,a7), axis=1)
A8
```

```python
a6 = np.ones((4,5), dtype = int)
a7 = np.zeros((3,5), dtype = int)
print(a6)
print(a7)

A8 = np.concatenate((a6,a7), axis=0)
A8
```

# Concatenation using: HSTACK & VSTACK

```python
a1 = np.ones((3,5), dtype = int)
a2 = np.zeros((3,5), dtype = int)

np.hstack((a1,a2))
```

```python
a1 = np.ones((3,5), dtype = int)
a2 = np.zeros((3,5), dtype = int)

np.vstack((a1,a2))
```

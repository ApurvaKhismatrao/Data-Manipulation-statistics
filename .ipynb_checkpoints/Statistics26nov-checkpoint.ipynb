# Probability

```python
# Probability Distribution: 
# a.uniform--equal probability to select every possibilities 
# b.binominal--
```

```python
import numpy as np
import pandas as pd
np.random.seed(42)   #random_state=42
coupons = np.random.randint(1,6,1000)
df = pd.DataFrame({"coupons":coupons})
df
```

```python
df.head()
```

```python
df.tail()
```

```python
df["coupons"].value_counts().sort_index()
```

```python
X = df["coupons"].value_counts().sort_index()
X
```

```python
Y = np.sort(df["coupons"].unique())
Y
```

```python
df["coupons"].values
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(4,4)) #figure size of the chart
plt.bar(x=X,height=Y)
plt.show()
```

# Visualization of Uniform Distribution

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(4,4)) #figure size of the chart
plt.barh(y=Y,width=X,edgecolor="black",color="purple")
plt.xlabel("counts of coupons")
plt.ylabel("coupons")
plt.grid()
plt.show()
```
